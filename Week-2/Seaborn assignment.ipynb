```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style='darkgrid')
```

##  Load Datasets

```python
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
flights = sns.load_dataset('flights')

tips.head()
```

##  Relational Plots

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', style='smoker')
plt.title('Total Bill vs Tip')
plt.show()
```

##  Categorical Plots

```python
sns.barplot(data=tips, x='day', y='tip')
plt.title('Average Tip by Day')
plt.show()
```

##  Distribution Plots

```python
sns.histplot(data=tips, x='total_bill', kde=True)
plt.title('Distribution of Total Bill')
plt.show()
```

##  Regression Plot

```python
sns.regplot(data=tips, x='total_bill', y='tip')
plt.title('Regression: Total Bill vs Tip')
plt.show()
```

```python
sns.set_theme(style="whitegrid")
sns.set_palette("Set2")

# New plot: Boxplot of tips by day
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='tip')
plt.title('Tip Distribution by Day')
plt.xlabel('Day')
plt.ylabel('Tip Amount')
plt.show()
```

**Insight:**  
The median tip is highest on Sunday at 3.15 dollars and lowest on Thursday at 2.30 dollars. Friday's median (3.00 dollars) is slightly higher than Saturday's (2.75 dollars), so tips are not higher on every weekend day.
